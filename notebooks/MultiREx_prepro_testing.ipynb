```python
import multirex as mrex

star   = mrex.Star(temperature=2566, radius=0.12, mass=0.0898)        # TRAPPIST-1-ish
planet = mrex.Planet(radius=0.92, mass=0.69)                          # TRAPPIST-1e-ish
atm    = mrex.Atmosphere(temperature=287, base_pressure=1e5, top_pressure=1e-3,
                         fill_gas="N2",
                         composition=dict(CO2=-2, CH4=-5, O3=-7, H2O=-3))  # log10 mixing ratios
planet.set_atmosphere(atm)

system = mrex.System(star=star, planet=planet, sma=0.029)             # sma in au
system.make_tm()

wn = mrex.Physics.wavenumber_grid(wl_min=0.6, wl_max=5.5, resolution=200)
bin_wn, depth = system.generate_spectrum(wn)   # depth = (Rp/Rs)^2 transit depth
```

```python
system = mrex.System(
    star=mrex.Star(temperature=(2500,2600), radius=(0.11,0.13), mass=(0.08,0.10)),
    planet=mrex.Planet(radius=(0.9,1.0), mass=(0.6,0.8),
        atmosphere=mrex.Atmosphere(
            temperature=(200,400), base_pressure=(1e5,1e5), top_pressure=(1e-3,1e-3),
            fill_gas="N2",
            composition=dict(CO2=-2, CH4=(-8,-1), O3=(-8,-1), H2O=(-8,-1)))),
    sma=(0.028,0.030))
system.make_tm()

results = system.explore_multiverse(
    wn, snr=5, n_universes=2000, n_observations=1,
    labels=["CH4","O3","H2O"],     # records presence + mixing ratios per row
    path="my_dataset", n_jobs=-1)  # writes parquet; n_jobs=-1 = all cores

df = results["observations"]
X  = df.data      # spectra (wavelength columns)
meta = df.params  # mixing ratios, SNR, system params  ← your regression targets live here
```
