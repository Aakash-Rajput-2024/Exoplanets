```python
!pip install tqdm
from tqdm import tqdm

# ================================================================
#  INARA Atmospheric Retrieval — Zorzan et al. (2025)
#  v2: Bulletproof column detection, handles dirty CSVs safely
# ================================================================

import os
import random
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
from scipy.stats import pearsonr

```

```python
# ================================================================
# 1. CONFIGURATION  — only change paths here
# ================================================================
SUMMARY_CSV = "D:\Exoplanets\Exoplanet2.O\summary.csv"
SPECTRA_DIR  = "D:\Exoplanets\Exoplanet2.O\Total_dataset"

MODEL_SAVE = os.path.join(
    MODEL_DIR,
    "best_model.pth"
)
PLOTS_DIR      = "plots"
CACHE_FILE = "cache/spectra_cache.npz"

SPECTRUM_LEN       = 4378
WAVELENGTH_CUTOFF  = 2.0   # µm  — keep only < 2µm per paper
TARGET_COLS        = ['H2O','CO2','O2','N2','CH4','N2O','CO','O3','SO2','NH3','C2H6','NO2']

SEED         = 42
BATCH_SIZE   = 64
NUM_EPOCHS   = 60
LR           = 1e-4
DROPOUT      = 0.2
PATIENCE     = 10
MC_SAMPLES   = 50
WEIGHT_DECAY = 1e-3

EPS_SPEC    = 1e-40        # for log10 of spectra  (values ~1e-23)
EPS_TARGET  = 1e-10        # for log10 of targets  (values down to 1e-10)umm 


# ================================================================
# 2. REPRODUCIBILITY
# ================================================================
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark     = False
# device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device: {device}")
```

```python
# ================================================================
# 3. SAFE CSV READER
#    Strips BOM + leading/trailing whitespace from ALL column names.
#    This is the root cause of the KeyError you saw.
# ================================================================
def safe_read_csv(fpath: str) -> pd.DataFrame:
    """Read CSV and clean column names — handles BOM, spaces, mixed case."""
    df = pd.read_csv(fpath, encoding='utf-8-sig')   # utf-8-sig strips BOM automatically
    df.columns = df.columns.str.strip()             # remove leading/trailing spaces
    return df

# ================================================================
# 4. FIND THE CORRECT COLUMN NAME (fuzzy match)
#    In case some files use slightly different naming.
# ================================================================
def find_column(df: pd.DataFrame, keyword: str) -> str:
    """
    Return the first column name that contains 'keyword' (case-insensitive).
    Raises a clear error if nothing matches.
    """
    keyword_lower = keyword.lower()
    matches = [c for c in df.columns if keyword_lower in c.lower()]
    if not matches:
        raise KeyError(
            f"No column containing '{keyword}' found.\n"
            f"Available columns: {df.columns.tolist()}"
        )
    return matches[0]   # take first match
```

```python
# ================================================================
# 5. BUILD FILE LOOKUP DICT  { planet_id_str : full_file_path }
# ================================================================
def build_file_dict(*spectra_dirs) -> dict:

    file_dict = {}

    for folder in spectra_dirs:

        if not os.path.isdir(folder):
            print(f"WARNING: spectra folder not found → {folder}")
            continue

        for fname in os.listdir(folder):

            if fname.endswith(".csv"):

                # safer filename extraction
                pid = os.path.splitext(fname)[0]

                # normalize
                pid = str(pid).strip().zfill(7)

                file_dict[pid] = os.path.join(folder, fname)

    print(f"Total spectrum files found: {len(file_dict)}")

    print("\nExample spectrum file IDs:")
    print(list(file_dict.keys())[:10])

    return file_dict

# ================================================================
# 6. LOAD SUMMARY FILES
# ================================================================
def load_summary(*csv_paths) -> pd.DataFrame:
    frames = []
    for p in csv_paths:
        if not os.path.exists(p):  
            print(f"  WARNING: summary file not found → {p}")
            continue
        df = safe_read_csv(p)
        print(f"  Loaded summary: {os.path.basename(p)}  ({len(df)} rows)")
        frames.append(df)

    combined = pd.concat(frames, ignore_index=True)

    # Normalise PlanetIndex to zero-padded 7-char string  e.g. "830" → "0000830"
    combined['planet_index'] = (
        combined['planet_index']
        .astype(str)
        .str.strip()
        .str.split('.')
        .str[0]
        .str.zfill(7)
    )
    combined = combined.drop_duplicates(subset='planet_index').reset_index(drop=True)
    print(f"Total unique planets in summary: {len(combined)}\n")
    return combined

# ================================================================
# 7. LOAD SPECTRA
# ================================================================
def load_spectra(summary_df: pd.DataFrame, file_dict: dict):
    X_list, y_list = [], []
    missing = skipped = 0

    for idx, row in tqdm(summary_df.iterrows(),
                         total= len(summary_df),
                         desc= "Loading spectra"):
        planet_id = row['planet_index']

        # --- find file ---
        if planet_id not in file_dict:
            missing += 1

            if missing < 20:
                print(f"Missing: {planet_id}")

            continue
        fpath = file_dict[planet_id]

        try:
            df = safe_read_csv(fpath)

            # --- find correct column names robustly ---
            wav_col    = find_column(df, 'wavelength')
            signal_col = find_column(df, 'planet_signal')

            # --- filter to < 2µm ---
            df = df[pd.to_numeric(df[wav_col], errors='coerce') < WAVELENGTH_CUTOFF]

            signal = pd.to_numeric(df[signal_col], errors='coerce').values
            signal = np.nan_to_num(signal.astype(np.float32), nan=0.0, posinf=0.0, neginf=0.0)

            # --- fixed length ---
            if len(signal) >= SPECTRUM_LEN:
                signal = signal[:SPECTRUM_LEN]
            else:
                signal = np.pad(signal, (0, SPECTRUM_LEN - len(signal)))

            X_list.append(signal)
            y_list.append([float(row[col]) for col in TARGET_COLS])

        except Exception as e:
            # Print first 5 errors then stay silent to avoid spam
            if skipped < 5:
                print(f"  SKIP {planet_id}: {e}")
            skipped += 1
            continue

    print(f"\nLoaded: {len(X_list)}  |  Missing files: {missing}  |  Bad files skipped: {skipped}")
    return np.array(X_list, dtype=np.float32), np.array(y_list, dtype=np.float32)
```

```python

# ================================================================
# 8. CNN MODEL  — exact paper architecture
# ================================================================
class MPSAdaptiveAvgPool1d(nn.Module):
    def __init__(self, output_size):
        super().__init__()
        self.output_size = output_size
        self.pool = nn.AdaptiveAvgPool1d(output_size)
    def forward(self, x):
        if x.device.type == "mps":
            rem = x.size(-1) % self.output_size
            if rem != 0:
                x = x[..., :-rem]
        return self.pool(x)

class INARA_CNN(nn.Module):
    def __init__(self, input_len: int, num_targets: int, dropout: float = 0.3):
        super().__init__()
        self.cnn = nn.Sequential(
            nn.Conv1d(1, 64, kernel_size=7, padding=3),
            nn.BatchNorm1d(64),
            nn.Tanh(),
            nn.MaxPool1d(2),
            
            # Layer 2: Detects specific molecular lines
            nn.Conv1d(64, 64, kernel_size=5, padding=2),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.MaxPool1d(2),
            
            # Layer 3: High-level features
            nn.Conv1d(64, 128, kernel_size=3, padding=1),
            nn.BatchNorm1d(128),
            nn.ReLU(),
            nn.MaxPool1d(2),

            #Layer 4
            nn.Conv1d(128, 256, kernel_size=3, padding=1),
            nn.BatchNorm1d(256),
            nn.ReLU(),
           
            MPSAdaptiveAvgPool1d(16)
        )
        self.fc = nn.Sequential(
        nn.Flatten(),

        nn.Linear(256 * 16, 1024),
        nn.ReLU(),
        nn.Dropout(dropout),

        nn.Linear(1024, 256),
        nn.ReLU(),
        nn.Dropout(dropout),

        nn.Linear(256, num_targets)
        )

    def forward(self, x):
        return self.fc(self.cnn(x))
# ================================================================
# 9. TRAIN
# ================================================================
def run_training(model, train_loader, val_loader):
    optimizer= torch.optim.Adam(model.parameters(), lr=LR)
    criterion= nn.MSELoss()
    best_val, counter = float('inf'), 0
    history = {'train': [], 'val': []}

    for epoch in range(1, NUM_EPOCHS + 1):

        # --- Training ---
        model.train()
        t_loss = 0.0

        train_bar = tqdm(
            train_loader,
            desc=f"Epoch {epoch}/{NUM_EPOCHS}"
        )

        for xb, yb in train_bar:
            xb, yb = xb.to(device), yb.to(device)
            optimizer.zero_grad()
            loss = criterion(model(xb), yb)
            loss.backward()
            optimizer.step()
            t_loss += loss.item()
           
            train_bar.set_postfix(
                loss=f"{loss.item():.6f}"
            )
        t_loss /= len(train_loader)

        # --- Validation ---
        model.eval()
        v_loss = 0.0
        with torch.no_grad():
            for xb, yb in val_loader:
                xb, yb = xb.to(device), yb.to(device)
                v_loss += criterion(model(xb), yb).item()
        v_loss /= len(val_loader)

        history['train'].append(t_loss)
        history['val'].append(v_loss)
        print(f"Epoch {epoch:03d}/{NUM_EPOCHS}  Train: {t_loss:.6f}  Val: {v_loss:.6f}")

        if v_loss < best_val:
            best_val, counter = v_loss, 0
            torch.save(model.state_dict(), MODEL_SAVE)
        else:
            counter += 1
            if counter >= PATIENCE:
                print(f"Early stopping at epoch {epoch}.")
                break

    return history

# ================================================================
# 10. MC DROPOUT INFERENCE
# ================================================================
def mc_predict(model, test_loader, scaler_y):
    model.train()   # dropout stays ON
    all_means, all_stds = [], []

    with torch.no_grad():
        for xb, _ in test_loader:
            xb   = xb.to(device)
            runs = np.stack([model(xb).detach().cpu().numpy() for _ in range(MC_SAMPLES)])
            all_means.append(runs.mean(axis=0))
            all_stds.append(runs.std(axis=0))

    means = scaler_y.inverse_transform(np.vstack(all_means))
    stds  = np.vstack(all_stds) * scaler_y.scale_
    return means, stds

# ================================================================
# 11. PLOTS
# ================================================================
def plot_loss(history):
    os.makedirs(PLOTS_DIR, exist_ok=True)
    plt.figure(figsize=(7, 4))
    plt.plot(history['train'], label='Train')
    plt.plot(history['val'],   label='Val')
    plt.xlabel('Epoch'); plt.ylabel('MSE Loss'); plt.title('Loss Curve')
    plt.legend(); plt.tight_layout()
    plt.savefig(f"{PLOTS_DIR}/loss_curve.png", dpi=120); plt.close()

def plot_true_vs_pred(y_true, y_pred):
    """Reproduces Fig 3 — log10 normalised abundance, predicted vs true."""
    os.makedirs(PLOTS_DIR, exist_ok=True)
    n_gases = len(TARGET_COLS)
    fig, axes = plt.subplots(3, 4, figsize=(16, 12))
    axes = axes.flatten()
 
    for i, gas in enumerate(TARGET_COLS):
        ax = axes[i]
        ax.scatter(y_true[:, i], y_pred[:, i], s=8, alpha=0.4, color='steelblue')
        lo = min(y_true[:, i].min(), y_pred[:, i].min())
        hi = max(y_true[:, i].max(), y_pred[:, i].max())
        ax.plot([lo, hi], [lo, hi], 'r--', lw=1.5, label='Perfect')
        r2 = r2_score(y_true[:, i], y_pred[:, i])
        ax.set_xlabel(f"True log₁₀({gas})", fontsize=8)
        ax.set_ylabel(f"Pred log₁₀({gas})", fontsize=8)
        ax.set_title(f"{gas}  R²={r2:.3f}", fontsize=9)
        ax.legend(fontsize=7)
 
    plt.suptitle("INARA: True vs Predicted (log₁₀ normalised abundance)", fontsize=12)
    plt.tight_layout()
    plt.savefig(f"{PLOTS_DIR}/fig3_all_gases.png", dpi=150, bbox_inches='tight')
    plt.close()
    print(f"Saved Fig3 → {PLOTS_DIR}/fig3_all_gases.png")

def plot_predictive_dist(mc_runs, y_true_row, planet_idx,
                         gas_a='H2O', gas_b='CH4'):
  
    os.makedirs(PLOTS_DIR, exist_ok=True)
    ia, ib = TARGET_COLS.index(gas_a), TARGET_COLS.index(gas_b)
    fig, axes = plt.subplots(1, 3, figsize=(14, 4))

    for ax, idx, gas, color in [(axes[0], ia, gas_a, 'steelblue'),
                                  (axes[1], ib, gas_b, 'darkorange')]:
        ax.hist(mc_runs[:, idx], bins=40, color=color, alpha=0.7)
        ax.axvline(y_true_row[idx], color='red',   lw=2, label='True')
        ax.axvline(mc_runs[:, idx].mean(), color='black', lw=2, ls='--', label='Pred median')
        ax.set_xlabel(gas); ax.set_title(f"{gas} distribution"); ax.legend()

    axes[2].scatter(mc_runs[:, ia], mc_runs[:, ib], s=5, alpha=0.4, color='purple')
    axes[2].axvline(y_true_row[ia], color='red', lw=1.5)
    axes[2].axhline(y_true_row[ib], color='red', lw=1.5)
    axes[2].scatter([y_true_row[ia]], [y_true_row[ib]],
                    color='red', s=100, zorder=5, marker='*', label='True')
    axes[2].set_xlabel(gas_a); axes[2].set_ylabel(f"log₁₀({gas_b})")
    axes[2].set_title(f"Joint {gas_a} & {gas_b}"); axes[2].legend()

    plt.suptitle(f"Planet #{planet_idx} — MC Predictive Distribution ({MC_SAMPLES} passes)")
    plt.tight_layout()
    out = f"{PLOTS_DIR}/fig4_planet{planet_idx}.png"
    plt.savefig(out, dpi=150, bbox_inches='tight'); plt.close()
    print(f"Saved Fig4 → {out}")
# ================================================================
# 12. MAIN
# ================================================================
def main():

   
    if len(X) == 0:
        raise RuntimeError(
            "No spectra loaded. Check your folder paths and file names."
        )

    print(f"X: {X.shape}   y: {y.shape}")

    # ---- Split: ~90% train, ~9% val, ~1% test ----
    X_train, X_tmp, y_train, y_tmp = train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=SEED
    )

    X_val, X_test, y_val, y_test = train_test_split(
        X_tmp,
        y_tmp,
        test_size=0.50,
        random_state=SEED
    )

    print(f"Train: {len(X_train)}  Val: {len(X_val)}  Test: {len(X_test)}")

    # ---- Scale ----
    sx, sy = StandardScaler(), StandardScaler()

    X_train = np.log10(X_train + 1e-30)
    X_val   = np.log10(X_val + 1e-30)
    X_test  = np.log10(X_test + 1e-30)

    X_train = X_train / (
        np.abs(X_train).max(axis=1, keepdims=True) + 1e-8
    )

    X_val = X_val / (
        np.abs(X_val).max(axis=1, keepdims=True) + 1e-8
    )

    X_test = X_test / (
        np.abs(X_test).max(axis=1, keepdims=True) + 1e-8
    )

    # ---- SCALE INPUT ----
    X_train = sx.fit_transform(X_train)
    X_val   = sx.transform(X_val)
    X_test  = sx.transform(X_test)

    # ---- SCALE TARGET ----
    y_train = sy.fit_transform(y_train)
    y_val   = sy.transform(y_val)
    y_test_sc = sy.transform(y_test)

    # ---- Tensors ----
    def tt(a):
        return torch.tensor(a, dtype=torch.float32)

    X_tr = tt(X_train).unsqueeze(1)
    X_va = tt(X_val).unsqueeze(1)
    X_te = tt(X_test).unsqueeze(1)

    y_tr = tt(y_train)
    y_va = tt(y_val)
    y_te = tt(y_test_sc)

    train_loader = DataLoader(
        TensorDataset(X_tr, y_tr),
        batch_size=BATCH_SIZE,
        shuffle=True
    )

    val_loader = DataLoader(
        TensorDataset(X_va, y_va),
        batch_size=BATCH_SIZE
    )

    test_loader = DataLoader(
        TensorDataset(X_te, y_te),
        batch_size=32
    )

    # ---- Model ----
    model = INARA_CNN(
        SPECTRUM_LEN,
        len(TARGET_COLS),
        DROPOUT
    ).to(device)

    params = sum(
        p.numel()
        for p in model.parameters()
        if p.requires_grad
    )

    print(f"\nTrainable parameters: {params:,}")

    # ---- Train ----
    print("\n=== Training ===")

    history = run_training(
        model,
        train_loader,
        val_loader
    )

    plot_loss(history)

    # ---- Inference ----
    print(f"\n=== MC Dropout Inference ({MC_SAMPLES} passes) ===")

    model.load_state_dict(torch.load(MODEL_SAVE))

    y_pred_sc, y_unc = mc_predict(
        model,
        test_loader,
        sy
    )

    # Back to log scale for Fig-3
    y_true_log = sy.inverse_transform(y_test_sc)
    y_pred_log = y_pred_sc

    # Back to abundance scale
    y_true_orig = 10 ** y_true_log
    y_pred_orig = 10 ** y_pred_log

    # ---- Metrics ----
    mse_orig = mean_squared_error(
        y_true_orig,
        y_pred_orig
    )

    print(f"\nTest MSE (abundance scale): {mse_orig:.4e}")

    print("\nPer-gas Metrics:")

    for i, gas in enumerate(TARGET_COLS):

        r2 = r2_score(
            y_true_orig[:, i],
            y_pred_orig[:, i]
        )

        corr, p_value = pearsonr(
            y_true_orig[:, i],
            y_pred_orig[:, i]
        )

        print(
            f"{gas:<6} | "
            f"R² = {r2:.4f} | "
            f"Correlation = {corr:.4f} | "
            f"p-value = {p_value:.4e}"
        )

    # ---- Plots ----
    plot_true_vs_pred(y_true_log, y_pred_log)

    # Fig 4
    idx = np.random.randint(0, len(X_te))

    model.train()

    with torch.no_grad():

        mc_runs = np.stack([
            model(X_te[idx:idx+1].to(device))
            .detach()
            .cpu()
            .numpy()[0]

            for _ in range(MC_SAMPLES)
        ])

    mc_runs_log = sy.inverse_transform(mc_runs)

    plot_predictive_dist(
        mc_runs_log,
        y_true_log[idx],
        planet_idx=idx
    )

    print(f"\nAll done. Plots saved in '{PLOTS_DIR}/'")
if __name__ == "__main__":
    main()
```
